```python
import numpy as np
import pandas as pd

# Load the dataset
df = pd.read_csv("googleplaystore.csv")

# 1. Handle 'Rating' column outlier and missing values
# Drop row index 10472 where columns are shifted (Rating = 19.0)
df = df.drop(df[df["Rating"] > 5].index)

# Fill missing 'Rating' values with the median
df["Rating"] = df["Rating"].fillna(df["Rating"].median())

# 2. Clean 'Reviews' column: Convert to numeric (integer)
df["Reviews"] = df["Reviews"].astype(int)


# 3. Clean 'Size' column: Convert to MB (float) and fill missing/variable sizes with median
def convert_size_to_mb(size):
    if pd.isna(size) or size == "Varies with device":
        return np.nan
    size = str(size).strip()
    if size.endswith("M"):
        return float(size[:-1])
    elif size.endswith("k") or size.endswith("K"):
        return float(size[:-1]) / 1024
    try:
        return float(size)
    except ValueError:
        return np.nan


df["Size"] = df["Size"].apply(convert_size_to_mb)
df["Size"] = df["Size"].fillna(df["Size"].median())

# 4. Clean 'Installs' column: Remove '+' and ',' and convert to numeric (integer)
df["Installs"] = (
    df["Installs"].str.replace("+", "", regex=False).str.replace(",", "", regex=False).astype(int)
)

# 5. Clean 'Price' column: Remove '$' and convert to numeric (float)
df["Price"] = df["Price"].str.replace("$", "", regex=False).astype(float)

# 6. Handle missing values for 'Type' and 'Content Rating' using mode
df["Type"] = df["Type"].fillna(df["Type"].mode()[0])
df["Content Rating"] = df["Content Rating"].fillna(df["Content Rating"].mode()[0])

# 7. Convert 'Last Updated' to datetime objects
df["Last Updated"] = pd.to_datetime(df["Last Updated"])
```

```python
# 1. Drop corrupted row (shifted columns where Rating > 5.0)
apps_df = apps_df[
    (apps_df['Rating'].isna()) | (apps_df['Rating'] <= 5.0)
].copy()

# 2. Clean Price: strip '$' and convert to float
apps_df['Price'] = (
    apps_df['Price'].astype(str).str.replace('$', '', regex=False).str.strip()
)
apps_df['Price'] = pd.to_numeric(apps_df['Price'], errors='coerce').fillna(0.0)

# 3. Impute missing Type based on Price
apps_df['Type'] = apps_df['Type'].fillna(
    apps_df['Price'].apply(lambda p: 'Free' if p == 0.0 else 'Paid')
)

# 4. Clean Reviews: convert to integer
apps_df['Reviews'] = pd.to_numeric(apps_df['Reviews'], errors='coerce').astype(
    'Int64'
)

# 5. Clean Installs: remove '+' and ',', parse to integer
apps_df['Installs'] = (
    apps_df['Installs']
    .astype(str)
    .str.replace('+', '', regex=False)
    .str.replace(',', '', regex=False)
    .str.strip()
)
apps_df['Installs'] = pd.to_numeric(apps_df['Installs'], errors='coerce').astype(
    'Int64'
)


# 6. Standardize Size into numeric Megabytes (MB)
def clean_size_mb(val):
  if pd.isna(val) or val == 'Varies with device':
    return np.nan
  val = str(val).strip()
  if val.endswith('M'):
    return float(val[:-1])
  elif val.endswith('k'):
    return float(val[:-1]) / 1024.0
  return np.nan


apps_df['Size_MB'] = apps_df['Size'].apply(clean_size_mb)

# 7. Convert Last Updated to Datetime
apps_df['Last Updated'] = pd.to_datetime(
    apps_df['Last Updated'], errors='coerce'
)

# 8. Deduplication: sort by Reviews descending and keep unique App names
apps_df = apps_df.drop_duplicates()
apps_clean = (
    apps_df.sort_values(by='Reviews', ascending=False)
    .drop_duplicates(subset=['App'], keep='first')
    .reset_index(drop=True)
)

print(
    f'Cleaned Apps shape: {apps_clean.shape} (from {len(apps_df)} initial rows)'
)
apps_clean.info()
```

```python
import pandas as pd

# 1. Load the user reviews dataset
reviews = pd.read_csv('googleplaystore_user_reviews.csv')

# 2. Drop rows with missing reviews
reviews_clean = reviews.dropna(subset=['Translated_Review']).copy()

# 3. Drop duplicate rows
reviews_clean = reviews_clean.drop_duplicates().reset_index(drop=True)

# Verify it exists
print("reviews_clean created successfully! Shape:", reviews_clean.shape)
```

```python
# 4. Sentiment Analysis & Common Themes in High vs. Low-Rated Apps
```

```python
from collections import Counter
import re

# Inner join valid reviews with cleaned app metadata
merged = reviews_clean.merge(
    apps_clean[['App', 'Rating', 'Category']], on='App', how='inner'
)

# Segment into high-rated (>= 4.5) and low-rated (< 3.5)
high_rated = merged[merged['Rating'] >= 4.5]
low_rated = merged[merged['Rating'] < 3.5]

# Compare sentiment polarity/proportions
print('Sentiment in High-Rated Apps:\n', high_rated['Sentiment'].value_counts(normalize=True))
print('\nSentiment in Low-Rated Apps:\n', low_rated['Sentiment'].value_counts(normalize=True))

# Keyword extraction (excluding common filler words)
stopwords = {'the', 'and', 'to', 'a', 'is', 'it', 'this', 'of', 'in', 'for', 'my', 'that', 'with', 'app'}
def get_top_terms(series, top_n=8):
  tokens = [
      word
      for text in series.dropna()
      for word in re.findall(r'\b[a-z]{3,}\b', text.lower())
      if word not in stopwords
  ]
  return Counter(tokens).most_common(top_n)

print('\nTop Keywords (High-Rated):', get_top_terms(high_rated['Translated_Review']))
print('Top Keywords (Low-Rated):', get_top_terms(low_rated['Translated_Review']))
```

```python
# Compute genre statistics with a minimum app count threshold
genre_ratings = (
    apps_clean.groupby('Genres')['Rating']
    .agg(['count', 'mean', 'median', 'std'])
    .query('count >= 30')
    .sort_values(by='mean', ascending=False)
)

print('Top 5 Genres by Rating:\n', genre_ratings.head(5).round(2))
print('\nBottom 5 Genres by Rating:\n', genre_ratings.tail(5).round(2))
```
